## YouTube Channels (Exploratory Data Analysis)

## Dataset Overview :-
This dataset contains information about the top YouTube channels worldwide.
It includes channel statistics such as:

- Subscribers
- Video views
- Upload counts
- Channel category
- Country
- Channel ranking
- Earnings and popularity metrics

The dataset helps analyze how successful YouTube creators and channels perform across different categories and countries.

## Objective:-
The objective of this project is to:
- Analyze top-performing YouTube channels
- Identify channels with highest subscribers and views
- Explore category-wise channel distribution
- Study country-wise YouTube growth
- Understand relationships between uploads, subscribers, and views
- Create meaningful visualizations using Python libraries

This project demonstrates how Exploratory Data Analysis (EDA) can be used to uncover trends and insights from social media platform data.

### Top 10 Channels by Video Views

### Channels with Highest Uploads

```python
top_uploads = df.sort_values(by='Video Uploads', ascending=False).head(10)

print(top_uploads[['Channel name', 'Video Uploads']])
```

```python
plt.figure(figsize=(10,5))
plt.bar(top_uploads['Channel name'], top_uploads['Video Uploads'])
plt.title("Top 5 Channels by Video Uploads",fontdict=font_title)
plt.xlabel("Channel Name",fontdict=font_labels)
plt.ylabel("Number of Video Uploads",fontdict=font_labels)
plt.xticks(rotation=45, ha='right')
plt.show()
```
