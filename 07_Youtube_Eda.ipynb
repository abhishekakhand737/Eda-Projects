### Data Cleaning

```python
df['Subscribers'] = (   df['Subscribers'].astype(str)    .str.replace(',', ''))
df['Video views'] = (    df['Video views'].astype(str) .str.replace(',', ''))
df['Video Uploads'] = ( df['Video Uploads'].astype(str).str.replace(',', ''))
df['Subscribers'] = pd.to_numeric(df['Subscribers'], errors='coerce')
df['Video views'] = pd.to_numeric(    df['Video views'],errors='coerce')
df['Video Uploads'] = pd.to_numeric(df['Video Uploads'],errors='coerce')
```

### Top 10 Channels by Subscribers

```python
top_subscribers = df.sort_values( by='Subscribers', ascending=False).head(10)
print(top_subscribers[    ['Channel name', 'Subscribers']])
```

```python
plt.figure(figsize=(14,6))
sns.barplot(y='Subscribers',  x='Channel name',  data=top_subscribers,color='teal')
plt.title("Top 10 Channels by Subscribers")
plt.xlabel("Channel Name")
plt.ylabel("Subscribers")
plt.xticks(rotation=50)
plt.show()
```

-  A small number of channels dominate the subscriber ecosystem.
- These channels typically represent global brands or creators.
- The gap between top channels and others is extremely large.
- Popularity on YouTube follows a power-law distribution.

### Grade Vs Number of Views

```python
grade_views = df.groupby('Grade')['Video views'].mean().sort_values(ascending=False)

print(grade_views)
```

```python
plt.figure(figsize=(10,5))
grade_views.plot(kind='bar',color='teal')
plt.title("Grade Vs Number of Views")
plt.xlabel("Grade")
plt.ylabel("Average Video Views")
plt.show()
```

- Channels with higher grades (A++, A+) record significantly higher average video views.
- This indicates that grade is a strong proxy for overall reach and audience engagement.
- Lower-grade channels attract fewer views, reflecting limited visibility or niche audiences.
- Video views emerge as a critical factor influencing channel grading and performance.

### Grade Vs Average Views per Video

```python
df['Average Views Per Video'] = (  df['Video views'] / df['Video Uploads'])
grade_avg_views = df.groupby('Grade')['Average Views Per Video'].mean().sort_values(ascending=False)

print(grade_avg_views)
```

```python
plt.figure(figsize=(10,5))
grade_avg_views.plot(kind='bar',color='teal')
plt.title("Grade Vs Average Views per Video")
plt.xlabel("Grade")
plt.ylabel("Average Views Per Video")
plt.show()
```

- Higher-grade channels (A++, A+) achieve significantly higher average views per video, indicating stronger audience engagement.
- Lower-grade channels show comparatively fewer average views, reflecting limited reach or engagement.
- This confirms that grade is closely associated with content quality, popularity, and audience loyalty.

### Grade with Higher Number of Video Uploads

```python
grade_uploads = df.groupby('Grade')['Video Uploads'].mean().sort_values(ascending=False)

print(grade_uploads)
```

```python
plt.figure(figsize=(10,5))
grade_uploads.plot(kind='bar',color='teal')
plt.title("Grade Vs Video Uploads")
plt.xlabel("Grade")
plt.ylabel("Average Video Uploads")
plt.show()
```

- Higher-grade channels (such as A++ and A+) generally show a higher average number of video uploads, reflecting consistent content creation.
- Lower-grade channels tend to upload fewer videos, indicating lower activity or shorter channel lifespan.
- This suggests that content consistency contributes to better channel grading, though it may not be the sole factor.
- Grade appears to capture a mix of activity level and overall channel performance.

### Grade Vs Number of Subscribers
