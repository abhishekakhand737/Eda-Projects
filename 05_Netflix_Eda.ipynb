## Netflix Shows & Movies (Exploratory Data Analysis)

This dataset contains information about movies and TV shows available on Netflix.
It includes details such as:

- Release Year

The analysis aims to uncover content trends, popular genres, country-wise production, and Netflix growth over the years.

### Fill Missing Values

```python
df['director'] = df['director'].fillna("Unknown")
df['cast'] = df['cast'].fillna("Not Available")
df['country'] = df['country'].fillna("Unknown")
df['rating'] = df['rating'].fillna("Not Rated")
```

### Movies vs TV Shows

```python
df['type'].value_counts()
```

```python
sns.countplot(x='type', data=df,color='teal')

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Count")

plt.show()
```

Netflix contains more Movies than TV Shows.

### Top 10 Content Producing Countries

```python
df['country'].value_counts().head(10)
```

```python
top_country = df['country'].value_counts().head(10)
top_country.plot(kind='bar', figsize=(8,5),color='teal')

plt.title("Top 10 Content Producing Countries")
plt.xlabel("Country")
plt.ylabel("Number of Contents")

plt.xticks(rotation=45)
plt.show()
```

United States and India contribute a large amount of Netflix content.

### Content Release Trend by Year

```python
df['release_year'].value_counts().head()
```

### Distribution of Movie Duration

```python
movies = df[df['type']=='Movie'].copy()
movies['duration'] = movies['duration'].str.replace(' min','')
movies['duration'] = pd.to_numeric(movies['duration'], errors='coerce')
```

```python
plt.figure(figsize=(8,5))
sns.histplot(movies['duration'], bins=30,color='teal')

plt.title("Movie Duration Distribution")
plt.xlabel("Duration in Minutes")
plt.ylabel("Frequency")

plt.show()
```

### Content Added Over Time

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Force to string type first, strip it, and parse the dates safely
df['date_added'] = pd.to_datetime(df['date_added'].astype(str).str.strip(), format='mixed')
df['year_added'] = df['date_added'].dt.year

# 2. Generate the plot
df['year_added'].value_counts().sort_index().plot(figsize=(9, 5))
plt.title("Content Added to Netflix Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Contents")
plt.show()
```

Netflix significantly expanded its content library after 2016.
