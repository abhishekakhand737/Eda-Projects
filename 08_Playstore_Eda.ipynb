### Data Cleaning

#### 1.Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

#### 2.Clean Installs Column

```python
df['Installs'] = (
    df['Installs']
    .astype(str)
    .str.replace(',', '')
    .str.replace('+', '')
)

df['Installs'] = pd.to_numeric(
    df['Installs'],
    errors='coerce'
)
```

#### 3.Clean Price Column

```python
df['Price'] = (df['Price'] .astype(str).str.replace('$', ''))
df['Price'] = pd.to_numeric(df['Price'],    errors='coerce')
```

#### 4.Convert Reviews to Numeric

```python
df['Reviews'] = pd.to_numeric(  df['Reviews'],
    errors='coerce')
```

#### 5.Convert Rating to Numeric

```python
df['Reviews'] = pd.to_numeric(  df['Reviews'],
    errors='coerce')
```

### App Category Distribution

```python
df['Category'].value_counts().head(10)
```

```python
plt.figure(figsize=(12,5))
df['Category'].value_counts().head(10).plot(kind='bar',color='violet')
plt.title("Top App Categories")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

### Top Rated Apps

```python
top_rated = df.sort_values(
    by='Rating',
    ascending=False
).head(10)

print(
    top_rated[
        ['App','Rating']
    ]
)
```

```python
plt.figure(figsize=(8,5))
sns.barplot(y='Rating', x='App',data=top_rated,color='violet')
plt.title("Top Rated Apps")
plt.xticks(rotation=60)
plt.show()
```

### Free vs Paid Apps

```python
df['Type'].value_counts()
```

```python
plt.figure(figsize=(4,4))
df['Type'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%')
plt.title("Free vs Paid Apps")
plt.ylabel("")
plt.show()
```

### Rating Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Rating'],    bins=20,color='teal')
plt.title("Rating Distribution")

plt.show()
```

### Category Vs Average Rating

```python
category_rating = df.groupby(
    'Category'
)['Rating'].mean().sort_values(
    ascending=False
).head(10)

print(category_rating)
```

```python
plt.figure(figsize=(12,5))
category_rating.plot(kind='bar',color='teal')
plt.title("Category Vs Average Rating")
plt.xlabel("Category")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.show()
```

### Average App Price

```python
df['Price'] = (df['Price'].astype(str).str.replace('$', '', regex=False))

df['Price'] = pd.to_numeric(df['Price'],errors='coerce')
print("Average App Price :",round(df['Price'].mean(),2))
```

### Correlation Heatmap

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    df[
        ['Rating','Reviews','Installs','Price']
    ].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```
