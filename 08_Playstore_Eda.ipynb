### Data Cleaning

#### Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

#### Clean Installs Column

```python
df['Installs'] = (
    df['Installs']
    .astype(str)
    .str.replace(',', '')
    .str.replace('+', '')
)

df['Installs'] = pd.to_numeric(
    df['Installs'],
    errors='coerce'
)
```

```python
df['Category'].value_counts().head(10)
```

```python
plt.figure(figsize=(12,5))

df['Category'].value_counts().head(10).plot(kind='bar')

plt.title("Top App Categories")

plt.xlabel("Category")

plt.ylabel("Count")

plt.show()
```

```python
Top Rated Apps
```

```python
top_rated = df.sort_values(
    by='Rating',
    ascending=False
).head(10)

print(
    top_rated[
        ['App','Rating']
    ]
)
```

```python
plt.figure(figsize=(12,5))

sns.barplot(
    x='Rating',
    y='App',
    data=top_rated
)

plt.title("Top Rated Apps")

plt.show()
```
