## IPL Match (Exploratory Data Analysis)

## Dataset Overview:- 
This IPL dataset is created by combining two important cricket datasets:
matches.csv → Contains match-level information
deliveries.csv → Contains ball-by-ball IPL data

After merging both datasets, the combined dataset provides detailed insights into:
- Teams
- Players
- Match results
- Batting performance
- Bowling statistics
- Toss decisions
- Venues
- Season-wise trends

The dataset includes ball-by-ball records from multiple IPL seasons, making it suitable for deep Exploratory Data Analysis (EDA) and cricket performance analysis.

## Objective:-
The objective of this IPL data analysis project is to:

- Analyze overall team performance in IPL history
- Identify top batsmen and bowlers
- Study winning patterns and toss impact
- Explore batting and bowling trends
- Discover venue-wise match insights
- Perform season-wise IPL analysis
- Understand player contributions using ball-by-ball data
- Create meaningful visualizations for cricket insights.

This project helps in understanding how teams and players have performed over different IPL seasons using data analysis and visualization techniques.

### Top 10 Batsmen by Total Runs

```python
top_batsmen = df.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(10)

print(top_batsmen)
```

```python
plt.figure(figsize=(10,5))

top_batsmen.plot(kind='bar',color= 'teal')

plt.title("Top 10 IPL Batsmen by Runs")
plt.xlabel("Batsman")
plt.ylabel("Total Runs")
plt.xticks(rotation=45)

plt.show()
```

Shows the highest run scorers in IPL history.

### Total Matches Played

```python
print("Total Matches Played :", df['match_id'].nunique())
```

### Total Teams Participated

```python
teams = pd.concat([df['team1'], df['team2']]).unique()
print("Total Teams Participated :", len(teams))
```

### Top 10 Players with Most Sixes

```python
top_sixes = df[df['batsman_runs'] == 6]['batter'].value_counts().head(10)
print(top_sixes)
```

```python
plt.figure(figsize=(10,5))

top_sixes.plot(kind='bar',color='teal')

plt.title("Players with Most Sixes")
plt.xlabel("Player")
plt.ylabel("Number of Sixes")
plt.xticks(rotation=45)

plt.show()
```

### Team with Highest Toss Win Percentage

```python
toss_percent = (df['toss_winner'].value_counts() / df['match_id'].nunique()) * 100
print(toss_percent.head())
```
