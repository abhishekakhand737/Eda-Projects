# 📊E-Commerce Exploratory Data Analysis 

### Dartaset Overview
This project performs Exploratory Data Analysis (EDA) on an e-commerce purchases dataset using Python, Pandas, Matplotlib, and Seaborn.

The dataset contains customer-level purchase information such as:
- Customer Email
- Job Title
- Credit Card Provider
- Purchase Price
- Language
- Purchase Timing (AM/PM)
- IP Address

The main goal of this analysis is to uncover customer behavior patterns, identify high-value users, and generate meaningful business insights.

### Highest, Lowest & Average Purchase Price

```python
print("Highest Purchase Price :", df['Purchase Price'].max())
print("Lowest Purchase Price :", df['Purchase Price'].min())
print("Average Purchase Price :", round(df['Purchase Price'].mean(), 2))
```

### Purchase Price Distribution

```python
plt.figure(figsize=(5,4))

sns.histplot(df['Purchase Price'], bins=20, kde=True , color='teal')

plt.title("Purchase Price Distribution")
plt.xlabel("Purchase Price")
plt.ylabel("Frequency")

plt.show()
```

The distribution shows customer spending patterns and common purchase ranges.
